# 1. Upload & Load Data

memuat seluruh sumber data mentah (SF, LPSE, dan SIJK) ke dalam lingkungan kerja sehingga dapat dianalisis dan diproses secara terintegrasi.

```python
import os
import re
import pandas as pd
import tkinter as tk
from tkinter import filedialog, messagebox

# 1) Atur keyword/pola deteksi di sini (sesuaikan dengan pola nama file Anda)
PATTERNS = {
    "SF": [
        r'(^|[^a-z0-9])sf([^a-z0-9]|$)',     # token "sf"
        r'(^|[^a-z0-9])s\.f([^a-z0-9]|$)',   # "s.f"
    ],
    "LPSE": [
        r'(^|[^a-z0-9])lpse([^a-z0-9]|$)',
    ],
    "SIJK": [
        r'(^|[^a-z0-9])sijk([^a-z0-9]|$)',
        r'(^|[^a-z0-9])si-jk([^a-z0-9]|$)',
    ],
}

def _normalize_filename(path: str) -> str:
    base = os.path.basename(path)
    name, _ = os.path.splitext(base)
    return name.lower()

def _detect_category(filename_lower_noext: str) -> str | None:
    hits = []
    for cat, patterns in PATTERNS.items():
        for pat in patterns:
            if re.search(pat, filename_lower_noext):
                hits.append(cat)
                break

    if not hits:
        return None
    if len(hits) > 1:
        raise ValueError(f"Ambigu: cocok ke beberapa kategori {hits}")
    return hits[0]

def _auto_map_files(paths: tuple[str, ...]) -> tuple[dict, list[str], list[str]]:
    mapping = {"SF": None, "LPSE": None, "SIJK": None}
    errors = []

    for p in paths:
        fname = _normalize_filename(p)

        try:
            cat = _detect_category(fname)
        except ValueError as e:
            errors.append(f"- {os.path.basename(p)}: {e}")
            continue

        if cat is None:
            errors.append(f"- {os.path.basename(p)}: tidak terdeteksi (keyword sf/lpse/sijk tidak ditemukan)")
            continue

        if mapping[cat] is not None:
            errors.append(f"- {os.path.basename(p)}: terdeteksi {cat}, tapi {cat} sudah terisi oleh {os.path.basename(mapping[cat])}")
            continue

        mapping[cat] = p

    missing = [k for k, v in mapping.items() if v is None]
    return mapping, errors, missing

def pilih_3_file_excel_tanpa_window_tk():

    root = tk.Tk()
    # Penting: withdraw agar jendela Tkinter kosong tidak muncul sama sekali.
    root.withdraw()
    # Supaya dialog muncul di depan (opsional tapi membantu)
    try:
        root.attributes("-topmost", True)
        root.update()
        root.attributes("-topmost", False)
    except Exception:
        pass

    while True:
        paths = filedialog.askopenfilenames(
            parent=root,
            title="Pilih file Excel (boleh acak). Auto-detect: SF, LPSE, SIJK dari nama file",
            filetypes=[("Excel files", "*.xlsx")]
        )

        if not paths:
            root.destroy()
            return None  # user cancel

        mapping, errors, missing = _auto_map_files(paths)

        if not errors and not missing:
            root.destroy()
            return mapping["SF"], mapping["LPSE"], mapping["SIJK"]

        # Buat pesan error ringkas tapi jelas
        msg_parts = []
        if missing:
            msg_parts.append("Kategori belum terisi: " + ", ".join(missing))
        if errors:
            msg_parts.append("\nDetail:\n" + "\n".join(errors))

        retry = messagebox.askretrycancel(
            "Mapping file gagal",
            "\n".join(msg_parts) + "\n\nKlik Retry untuk memilih ulang, atau Cancel untuk batal.",
            parent=root
        )

        if not retry:
            root.destroy()
            return None


# ====== Pemakaian ======
result = pilih_3_file_excel_tanpa_window_tk()
if not result:
    raise SystemExit("Proses dibatalkan atau file belum berhasil terdeteksi lengkap.")

sf_file, lpse_file, sijk_file = result

df_sf = pd.read_excel(sf_file, dtype=str)
df_lpse = pd.read_excel(lpse_file, dtype=str)
df_sijk = pd.read_excel(sijk_file, dtype=str)

# Display aman di Jupyter / non-Jupyter
try:
    from IPython.display import display
except Exception:
    def display(x):
        print(x)

print("=== Preview 5 baris File SF ===")
display(df_sf.head())

print("\n=== Preview 5 baris File LPSE ===")
display(df_lpse.head())

print("\n=== Preview 5 baris File SIJK ===")
display(df_sijk.head())

display(pd.DataFrame({
    "File": ["SF", "LPSE", "SIJK"],
    "Jumlah_Baris": [len(df_sf), len(df_lpse), len(df_sijk)],
    "Jumlah_Kolom": [df_sf.shape[1], df_lpse.shape[1], df_sijk.shape[1]]
}))

print("Data berhasil dimuat.")
```

Hasil output memastikan seluruh file berhasil dibaca ke dalam DataFrame sehingga siap digunakan pada tahapan analisis selanjutnya.

# 2. Eksplorasi Data Awal (EDA)

Eksplorasi awal diperlukan untuk memahami kondisi data mentah, terutama ukuran data dan tingkat kelengkapan nilai, sebelum dilakukan transformasi lebih lanjut.

```python

import pandas as pd
from IPython.display import display

datasets = {
    "SF": df_sf,
    "LPSE": df_lpse,
    "SIJK": df_sijk
}

ringkasan_per_file = []

total_baris = 0
total_kolom = 0
total_missing = 0
total_terisi = 0
total_sel = 0

for nama, df in datasets.items():
    jumlah_baris = df.shape[0]
    jumlah_kolom = df.shape[1]

    total_cell = jumlah_baris * jumlah_kolom
    missing = int(df.isna().sum().sum())
    terisi = int(total_cell - missing)

    persen_missing = round((missing / total_cell) * 100, 2) if total_cell > 0 else 0
    persen_terisi = round((terisi / total_cell) * 100, 2) if total_cell > 0 else 0

    ringkasan_per_file.append({
        "File": nama,
        "Jumlah_Baris": jumlah_baris,
        "Jumlah_Kolom": jumlah_kolom,
        "Jumlah_Nilai_Terisi": terisi,
        "Jumlah_Missing_Value": missing,
        "Persen_Nilai_Terisi (%)": persen_terisi,
        "Persen_Missing_Value (%)": persen_missing
    })

    total_baris += jumlah_baris
    total_kolom += jumlah_kolom
    total_missing += missing
    total_terisi += terisi
    total_sel += total_cell

# Ringkasan per file
print("=== RINGKASAN DATA PER FILE ===")
display(pd.DataFrame(ringkasan_per_file))
print()
print()

# Ringkasan gabungan semua file
ringkasan_total = pd.DataFrame([{
    "Total_Baris_Semua_File": total_baris,
    "Total_Kolom_Semua_File": total_kolom,
    "Total_Nilai_Terisi_Semua_File": total_terisi,
    "Total_Missing_Value_Semua_File": total_missing,
    "Persen_Nilai_Terisi_Semua_File (%)": round((total_terisi / total_sel) * 100, 2) if total_sel > 0 else 0,
    "Persen_Missing_Value_Semua_File (%)": round((total_missing / total_sel) * 100, 2) if total_sel > 0 else 0
}])

print("=== RINGKASAN GABUNGAN SEMUA FILE ===")
display(ringkasan_total)
```

EDA memberikan gambaran awal ukuran data dan proporsi missing value pada masing-masing file sehingga potensi masalah data dapat diantisipasi sejak awal.

```python
# FIX2 - STEP 1: DROP DUPLIKAT LPSE BERDASARKAN kd_penyedia

def row_completeness_score(row):
    # hitung jumlah kolom yang terisi (tidak kosong / tidak NaN)
    return row.replace("", np.nan).notna().sum()

if "kd_penyedia" in df_lpse.columns:

    before = len(df_lpse)

    df_lpse["_kelengkapan"] = df_lpse.apply(row_completeness_score, axis=1)

    # urutkan: kd_penyedia, kelengkapan desc
    df_lpse = (
        df_lpse
        .sort_values(by=["kd_penyedia", "_kelengkapan"], ascending=[True, False])
        .drop_duplicates(subset=["kd_penyedia"], keep="first")
        .drop(columns=["_kelengkapan"])
        .reset_index(drop=True)
    )

    display(pd.DataFrame({
        "LPSE_Sebelum": [before],
        "LPSE_Sesudah": [len(df_lpse)],
        "Duplikat_kd_penyedia_Dihapus": [before - len(df_lpse)]
    }))

    print("Dedup LPSE berdasarkan kd_penyedia (dipilih yang paling lengkap) selesai.\n")

else:
    print("⚠️ Kolom 'kd_penyedia' tidak ditemukan di LPSE, dedup tidak dijalankan.\n")
```

# 4. Mapping Wilayah LPSE & SIJK ke Skema SF

"kdprov" dan "kd_kab" adalah variabel utama deduplikasi dan harus diseragamkan sebelum RapidFuzz dijalankan.

```python
import pandas as pd
import numpy as np
import re
from rapidfuzz import fuzz
from datetime import datetime
from IPython.display import display
from openpyxl import load_workbook
from openpyxl.styles import PatternFill
# from google.colab import files

def clean_text(x):
    x = str(x).lower()
    x = re.sub(r"[^a-z0-9 ]", " ", x)
    x = re.sub(r"\b(pt|cv|tbk|persero|ud|firma)\b", " ", x)
    return re.sub(r"\s+", " ", x).strip()

TAHUN_REVISI = datetime.now().year
print("Setup & normalisasi siap\n")

TAHUN_REVISI = datetime.now().year

map_kualifikasi = {
    "spesialis": "1", "kecil": "2",
    "menengah": "3", "besar": "4",
    "tidak memiliki sbu aktif": "9"
}
map_skala = {"kecil": "2", "menengah": "3", "besar": "4"}
map_badan_usaha = {
    "pt. persero (bumn/bumd)": "1",
    "pt": "2", "cv": "3",
    "koperasi": "4", "kantor perwakilan bujka": "5"
}

def get_and_map_column(df, source_col, map_dict, default_val=""):
    if source_col in df.columns:
        column_data = df[source_col]

        if isinstance(column_data, pd.DataFrame):
            series_to_process = column_data.iloc[:, 0]
        else:
            series_to_process = column_data

        return (
            series_to_process
            .astype(str)
            .str.lower()
            .map(map_dict)
            .fillna(default_val)
        )
    else:
        return pd.Series([default_val] * len(df), index=df.index)

def standardize_lpse(df):
    df = df.copy().reset_index(drop=True)
    df = df.rename(columns={
        "nama_penyedia": "nama_perusahaan",
        "npwp_penyedia": "npwp",
        "alamat": "alamat_perusahaan",
        "telepon": "no_telp",
        "nomor_izin_usaha": "nib",
        "bentuk_usaha": "badan_usaha",
        "kualifikasi_usaha": "kualifikasi",
        "kd_penyedia": "kd_penyedia",
        "kdprov": "prov",
        "kd_kab": "kab",
        "nmkab": "nm_kab",
        "nmprov": "nm_prov"
    })
    df["badan_usaha"] = get_and_map_column(df, "badan_usaha", map_badan_usaha, "")
    df["kualifikasi"] = get_and_map_column(df, "kualifikasi", map_kualifikasi, "")
    df["tahun_revisi"] = str(TAHUN_REVISI)
    df["kategori"] = "F"
    df["sumber_data"] = "1"
    return df

def standardize_sijk(df):
    df = df.copy()
    df = df.loc[:, ~df.columns.duplicated()].reset_index(drop=True)
    df = df.rename(columns={
        "nama_bu": "nama_perusahaan",
        "npwp_bu": "npwp",
        "alamat_bu": "alamat_perusahaan",
        "bentuk_usaha_bu": "bentuk_usaha_sijk",
        "kualifikasi_bu": "kualifikasi_sijk",
        "skala_usaha": "skala_usaha_sijk",
        "telepon_bu": "no_telp",
        "email_bu": "email",
        "sub_klasifikasi": "pekerjaan_utama",
        "kdprov": "prov",
        "kd_kab": "kab",
        "nmkab": "nm_kab",
        "nmprov": "nm_prov"
    })
    df["badan_usaha"] = get_and_map_column(df, "bentuk_usaha_sijk", map_badan_usaha, "")
    df["kualifikasi"] = get_and_map_column(df, "kualifikasi_sijk", map_kualifikasi, "")
    df["skala_usaha"] = get_and_map_column(df, "skala_usaha_sijk", map_skala, "")
    df["tahun_revisi"] = str(TAHUN_REVISI)
    df["kategori"] = "F"
    df["sumber_data"] = "2"
    return df

df_lpse = standardize_lpse(df_lpse)
df_sijk = standardize_sijk(df_sijk)

display(pd.DataFrame({
    "File": ["LPSE","SIJK"],
    "Baris": [len(df_lpse), len(df_sijk)],
    "Kolom": [df_lpse.shape[1], df_sijk.shape[1]]
}))
print("Standarisasi LPSE & SIJK selesai\n")
print()
print()

# FUNGSI DEDUP FUZZY 4 KOLOM + MERGE ISI
def dedup_fuzzy_4kolom_flag(df):
    df = df.copy().reset_index(drop=True)

    df["nama_clean"]   = df["nama_perusahaan"].apply(clean_text)
    df["alamat_clean"] = df["alamat_perusahaan"].apply(clean_text)
    df["flag_dup_fuzzy"] = "UNIK"
    df["dup_group_id"]   = ""
    df["sim_nama"]       = np.nan
    df["sim_alamat"]     = np.nan

    group_id = 1

    for i in range(len(df)):
        if df.at[i,"dup_group_id"] != "":
            continue

        df.at[i,"dup_group_id"] = f"G{group_id}"

        for j in range(i+1, len(df)):
            sim_nama   = fuzz.token_set_ratio(df.at[i,"nama_clean"], df.at[j,"nama_clean"])
            sim_alamat = fuzz.token_set_ratio(df.at[i,"alamat_clean"], df.at[j,"alamat_clean"])

            prov_i = "" if pd.isna(df.at[i,"prov"]) else str(df.at[i,"prov"]).strip()
            prov_j = "" if pd.isna(df.at[j,"prov"]) else str(df.at[j,"prov"]).strip()
            kab_i  = "" if pd.isna(df.at[i,"kab"])  else str(df.at[i,"kab"]).strip()
            kab_j  = "" if pd.isna(df.at[j,"kab"])  else str(df.at[j,"kab"]).strip()

            if (
                sim_nama >= 80 and
                sim_alamat >= 80 and
                ((prov_i == prov_j) or (prov_i == "") or (prov_j == "")) and
                ((kab_i  == kab_j)  or (kab_i  == "") or (kab_j  == ""))
            ):
                df.at[j,"flag_dup_fuzzy"] = "DUPLIKAT"
                df.at[j,"dup_group_id"]   = f"G{group_id}"
                df.at[j,"sim_nama"]       = sim_nama
                df.at[j,"sim_alamat"]     = sim_alamat

        group_id += 1

    return df
```

```python
kolom_std = pd.DataFrame({
    "SIJK": [
        "nama_bu",
        "npwp_bu",
        "alamat_bu",
        "bentuk_usaha_bu",
        "kualifikasi_bu",
        "kdprov",
        "kd_kab",
        "nmkab",
        "nmprov"
    ],
    "LPSE": [
        "nama_penyedia",
        "npwp_penyedia",
        "alamat",
        "bentuk_usaha",
        "kualifikasi_usaha",
        "kdprov",
        "kd_kab",
        "nmkab ",
        "nmprov"
    ],
    "SF": [
        "nama_perusahaan",
        "npwp",
        "alamat_perusahaan",
        "badan_usaha",
        "kualifikasi",
        "prov",
        "kab",
        "nm_kab",
        "nm_prov"
    ]
})
display(kolom_std)
print("Standarisasi kolom siap\n")
```

```python
# 1. DEDUP INTERNAL SIJK
df_sijk_flag = dedup_fuzzy_4kolom_flag(df_sijk)

display(pd.DataFrame({
    "SIJK_Total": [len(df_sijk_flag)],
    "SIJK_Duplikat": [(df_sijk_flag["flag_dup_fuzzy"] == "DUPLIKAT").sum()]
}))
print("Dedup internal SIJK selesai (flag)\n")
```

# 7. RapidFuzz Tahap 3 — Deduplikasi Gabungan SIJK + LPSE

Menghapus duplikasi lintas sumber sebelum data digabung ke SF.

```python
# 3. DEDUP SILANG LPSE + SIJK
lpse_sijk = pd.concat([df_lpse_flag, df_sijk_flag], ignore_index=True)
lpse_sijk_flag = dedup_fuzzy_4kolom_flag(lpse_sijk)

display(pd.DataFrame({
    "Gabungan_Total": [len(lpse_sijk_flag)],
    "Gabungan_Duplikat": [(lpse_sijk_flag["flag_dup_fuzzy"] == "DUPLIKAT").sum()]
}))
print("Dedup silang LPSE–SIJK selesai (flag)\n")
print()
print()

sf_cols = list(df_sf.columns)
extra_cols = [c for c in lpse_sijk_flag.columns if c not in sf_cols]
final_cols = sf_cols + extra_cols

def reindex(df):
    df = df.copy()
    for c in final_cols:
        if c not in df.columns:
            df[c] = ""
    return df[final_cols]

df_final = pd.concat([
    reindex(df_sf),
    reindex(lpse_sijk_flag)
], ignore_index=True)

display(pd.DataFrame({
    "Total_Baris_Akhir": [len(df_final)],
    "Total_Kolom_Akhir": [df_final.shape[1]],
    "Total_Duplikat": [(df_final["flag_dup_fuzzy"]=="DUPLIKAT").sum()]
}))

print("Dataset final FIX2 terbentuk.")
print()
print()

display(pd.DataFrame({
    "SF_Baris": [len(df_sf)],
    "LPSE_SIJK_Baris": [len(lpse_sijk_flag)],
    "TOTAL_FINAL": [len(df_final)],
    "TOTAL_KOLOM": [df_final.shape[1]],
    "TOTAL_DUPlIKAT_FINAL": [(df_final["flag_dup_fuzzy"]=="DUPLIKAT").sum()]
}))
print("Penggabungan final selesai (SF tidak diubah)\n")
```

Data eksternal kini bebas duplikasi lintas sumber.
